 Import the dataset using Pandas.

```python
import pandas as pd

df = pd.read_csv('../../data/raw/Dataset.csv')
```

```python
print(df['release_year'].value_counts())
```

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce',format='mixed')
print(df['date_added'].value_counts())
```

```python
print(df['country'].value_counts())
```

```python
df['country']=df['country'].replace('Not Given','Unknown')
print(df['country'].value_counts())
```

```python
country_code={
    'United States':'US',
    'India':'IN',
    'United Kingdom':'UK',
    'Pakistan':'PK',
    'Iran':'IR',
    'West Germany':'DE',
    'Greece':'GR',
    'Zimbabwe':'ZW',
    'Soviet Union':'SU',
    'France':'FR',
    'Germany':'DE',
    'Namibia':'NA',
    'Mozambique':'MZ',
    'Brazil':'BR'
}
df['country']=df['country'].replace(country_code)
print(df['country'].value_counts())
```

```python
df.head(10)
```

```python

extracted = df['duration'].str.extract(r'(\d+)\s*([a-zA-Z]+)')
# creation of news columns
df['duration_value'] = pd.to_numeric(extracted[0], errors='coerce')
df['duration_unit'] = extracted[1].str.lower()

# 3. Normalization
unit_mapping = {
    'min': 'minutes', 'minute': 'minutes', 'minutes': 'minutes',
    'h': 'hours', 'hour': 'hours', 'hours': 'hours', 'hr': 'hours',
    'season': 'seasons', 'seasons': 'seasons'
}
df['duration_unit'] = df['duration_unit'].map(unit_mapping)
```
